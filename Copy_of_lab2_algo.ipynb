```python
def read_data_from_file(filename):
    """Reads integers from a file and returns them as a list."""
    try:
        with open(filename, 'r') as file:
            return [int(line.strip()) for line in file if line.strip().isdigit()]
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return []

def linear_search(data, target):
    """Performs linear search. Returns the index if found, else -1."""
    for index, value in enumerate(data):
        if value == target:
            return index
    return -1

def search_in_file(filename, target, label=""):
    """Helper to search in a specific file and display results."""
    data = read_data_from_file(filename)
    if not data:
        return
    index = linear_search(data, target)
    if index != -1:
        print(f"{target} found in {label} data at index {index}.")
    else:
        print(f"{target} not found in {label} data.")

def main():
    # File names
    sorted_file = "SortedData5k.txt"
    unsorted_file = "UnsortedData5k.txt"

    # User input
    try:
        target = int(input("Enter the number to search for: "))
    except ValueError:
        print("Invalid input. Please enter a number.")
        return

    # Search in both files
    print("\nSearching using Linear Search...")
    search_in_file(unsorted_file, target, "unsorted")
    search_in_file(sorted_file, target, "sorted")

if __name__ == "__main__":
    main()
```

```python
def read_data_from_file(filename):
    """Reads integers from a file and returns them as a list."""
    try:
        with open(filename, 'r') as file:
            return [int(line.strip()) for line in file if line.strip().isdigit()]
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return []

def binary_search(data, target):
    """Performs binary search on sorted data. Returns index if found, else -1."""
    left, right = 0, len(data) - 1
    while left <= right:
        mid = (left + right) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

def main():
    # File with sorted data
    sorted_file = "SortedData5k.txt"

    # Get user input
    try:
        target = int(input("Enter the number to search for: "))
    except ValueError:
        print("Invalid input. Please enter a number.")
        return

    # Read sorted data
    data = read_data_from_file(sorted_file)
    if not data:
        return

    # Binary search
    print("\nSearching using Binary Search...")
    index = binary_search(data, target)
    if index != -1:
        print(f"{target} found at index {index} in sorted data.")
    else:
        print(f"{target} not found in sorted data.")

if __name__ == "__main__":
    main()
```

```python
import time

def read_data_from_file(filename):
    """Reads integers from a file and returns them as a list."""
    try:
        with open(filename, 'r') as file:
            return [int(line.strip()) for line in file if line.strip().isdigit()]
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return []

def linear_search(data, target):
    """Performs linear search. Returns the index if found, else -1."""
    for index, value in enumerate(data):
        if value == target:
            return index
    return -1

def binary_search(data, target):
    """Performs binary search. Returns index if found, else -1."""
    left, right = 0, len(data) - 1
    while left <= right:
        mid = (left + right) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

def main():
    sorted_file = "SortedData5k.txt"
    unsorted_file = "UnsortedData5k.txt"

    try:
        target = int(input("Enter the number to search for: "))
    except ValueError:
        print("Invalid input. Please enter a valid number.")
        return

    # Linear Search on Unsorted Data
    unsorted_data = read_data_from_file(unsorted_file)
    if unsorted_data:
        print("\n--- Linear Search on Unsorted Data ---")
        start_time = time.time()
        index = linear_search(unsorted_data, target)
        end_time = time.time()
        if index != -1:
            print(f"{target} found at index {index}.")
        else:
            print(f"{target} not found.")
        print(f"Time taken: {(end_time - start_time):.6f} seconds")

    # Linear Search on Sorted Data
    sorted_data = read_data_from_file(sorted_file)
    if sorted_data:
        print("\n--- Linear Search on Sorted Data ---")
        start_time = time.time()
        index = linear_search(sorted_data, target)
        end_time = time.time()
        if index != -1:
            print(f"{target} found at index {index}.")
        else:
            print(f"{target} not found.")
        print(f"Time taken: {(end_time - start_time):.6f} seconds")

        # Binary Search on Sorted Data
        print("\n--- Binary Search on Sorted Data ---")
        start_time = time.time()
        index = binary_search(sorted_data, target)
        end_time = time.time()
        if index != -1:
            print(f"{target} found at index {index}.")
        else:
            print(f"{target} not found.")
        print(f"Time taken: {(end_time - start_time):.6f} seconds")

if __name__ == "__main__":
    main()
```
